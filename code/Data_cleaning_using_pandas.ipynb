### Standardizing and normalizing the data in gender column

```python
# Gender columns should contain only few categories
# Checking for abnormal data
df['gender'].unique() 
```

```python
# Changing all Male and Female to proper case, others to other and filling nan with unknown
# all Male gender to "Male"
df['gender'] = df['gender'].replace(['MALE','male','M','m'],'Male')
```

```python
df['gender'].unique()
```

```python
# all Female gender to "Female"
df['gender'] = df['gender'].replace(['FEMALE','female','F','f'],'Female')
```

```python
# changing others to "Other" 
df['gender'] = df['gender'].replace(['Others','others','other'],'Other')
```

```python
# changing everything else to "Unknown"
df['gender'] = df['gender'].replace([' ','unknown'],'Unknown')
```

```python
# finally filling nan with "Unknown" as well
df['gender'] = df['gender'].fillna('Unknown')
```

```python
df['gender'].unique()
```
